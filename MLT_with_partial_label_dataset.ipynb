```python
# -*- coding: utf-8 -*-

"""
RICE LEAF DISEASE
STL vs FULL MTL vs PARTIAL-LABEL MTL

Tasks:
    1. Disease Classification
    2. Disease-Lesion Segmentation

Experiments:
    E1  : STL Classification (100%)
    E2  : STL Segmentation   (100%)
    E3  : Full-label MTL     (100% + 100%)

    E4  : Partial MTL 50/50
    E5  : STL Classification 50%
    E6  : STL Segmentation   50%

    E7  : Partial MTL 70/30
    E8  : STL Classification 70%
    E9  : STL Segmentation   30%

    E10 : Partial MTL 30/70
    E11 : STL Classification 30%
    E12 : STL Segmentation   70%

Important:
    Partial MTL uses the SAME images but hides one task's supervision.
    Original image/mask files are NEVER modified.

Evaluation:
    Classification -> Accuracy, Macro-F1
    Segmentation   -> Dice, IoU

Split:
    70% Train
    15% Validation
    15% Test
"""


# ============================================================
# 1. INSTALLATION
# ============================================================

!pip install -q segmentation-models-pytorch


# ============================================================
# 2. IMPORTS
# ============================================================

import os
import json
import random
import copy
import subprocess
import time

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader, Subset

import torchvision.transforms as T
import torchvision.transforms.functional as TF

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score

from tqdm.auto import tqdm

import segmentation_models_pytorch as smp


# ============================================================
# 3. MOUNT GOOGLE DRIVE
# ============================================================

from google.colab import drive

drive.mount("/content/drive")


# ============================================================
# 4. PATH CONFIGURATION
# ============================================================

PROJECT_DIR = "/content/drive/MyDrive/Rice_Disease_Project"

IMAGE_ZIP = os.path.join(
    PROJECT_DIR,
    "Rice Leaf Disease Image Samples.zip"
)

MASK_ZIP = os.path.join(
    PROJECT_DIR,
    "Rice_Leaf_Disease_Segmentation_Masks.zip"
)

DATASET_DIR = "/content/dataset"

IMAGE_ROOT = os.path.join(
    DATASET_DIR,
    "images"
)

MASK_ROOT = os.path.join(
    DATASET_DIR,
    "masks"
)

IMAGE_7Z = os.path.join(
    IMAGE_ROOT,
    "Rice Leaf Disease Image Samples",
    "Rice Leaf Disease Images.7z"
)

EXTRACTED_IMAGE_DIR = os.path.join(
    IMAGE_ROOT,
    "extracted_images"
)

IMAGE_DIR = os.path.join(
    EXTRACTED_IMAGE_DIR,
    "Rice Leaf Disease Images"
)

MASK_DIR = os.path.join(
    MASK_ROOT,
    "mask-vesrion1"
)

SAVE_DIR = os.path.join(
    PROJECT_DIR,
    "saved_models",
    "partial_label_mtl_experiment"
)

RESULTS_DIR = os.path.join(
    PROJECT_DIR,
    "experiment_results"
)

os.makedirs(IMAGE_ROOT, exist_ok=True)
os.makedirs(MASK_ROOT, exist_ok=True)
os.makedirs(EXTRACTED_IMAGE_DIR, exist_ok=True)
os.makedirs(SAVE_DIR, exist_ok=True)
os.makedirs(RESULTS_DIR, exist_ok=True)

print("=" * 80)
print("PROJECT PATHS")
print("=" * 80)

print("Project :", PROJECT_DIR)
print("Images  :", IMAGE_DIR)
print("Masks   :", MASK_DIR)
print("Models  :", SAVE_DIR)
print("Results :", RESULTS_DIR)


# ============================================================
# 5. EXTRACT DATASET
# ============================================================

# ------------------------------------------------------------
# 5.1 Extract image ZIP
# ------------------------------------------------------------

if not os.path.exists(IMAGE_7Z):

    print("\n📦 Extracting image ZIP...")

    !unzip -q "$IMAGE_ZIP" -d "$IMAGE_ROOT"

    print("✅ Image ZIP extracted.")

else:

    print("\n✅ Image ZIP already extracted.")


# ------------------------------------------------------------
# 5.2 Extract mask ZIP
# ------------------------------------------------------------

if not os.path.exists(MASK_DIR):

    print("\n📦 Extracting mask ZIP...")

    !unzip -q "$MASK_ZIP" -d "$MASK_ROOT"

    print("✅ Mask ZIP extracted.")

else:

    print("\n✅ Mask dataset already extracted.")


# ------------------------------------------------------------
# 5.3 Install 7-Zip if necessary
# ------------------------------------------------------------

if not os.path.exists(IMAGE_7Z):

    raise FileNotFoundError(
        f"\nCould not find:\n{IMAGE_7Z}\n"
        "Please check the image ZIP structure."
    )


result = subprocess.run(
    ["bash", "-c", "command -v 7z"],
    capture_output=True,
    text=True
)

if result.returncode != 0:

    print("\n📦 Installing 7-Zip...")

    !apt-get update -qq
    !apt-get install -y -qq p7zip-full

else:

    print("✅ 7-Zip already installed.")


# ------------------------------------------------------------
# 5.4 Extract .7z images
# ------------------------------------------------------------

if not os.path.exists(IMAGE_DIR):

    print("\n📦 Extracting image .7z archive...")

    !7z x "$IMAGE_7Z" -o"$EXTRACTED_IMAGE_DIR" -y

    print("✅ Image extraction complete.")

else:

    print("✅ Images already extracted.")


# ============================================================
# 6. DATASET STRUCTURE VERIFICATION
# ============================================================

CLASSES = [
    "Blast",
    "Tungro",
    "Brownspot",
    "Bacterialblight"
]

CLASS_MAP = {
    "blast": 0,
    "tungro": 1,
    "brownspot": 2,
    "bacterialblight": 3
}

CLASS_NAMES = [
    "Blast",
    "Tungro",
    "Brownspot",
    "Bacterialblight"
]

NUM_CLASSES = 4


print("\n" + "=" * 80)
print("DATASET STRUCTURE CHECK")
print("=" * 80)

total_images = 0
total_masks = 0

for cls in CLASSES:

    image_class_dir = os.path.join(
        IMAGE_DIR,
        cls
    )

    mask_class_dir = os.path.join(
        MASK_DIR,
        cls
    )

    if not os.path.isdir(image_class_dir):
        raise FileNotFoundError(
            f"Missing image directory:\n{image_class_dir}"
        )

    if not os.path.isdir(mask_class_dir):
        raise FileNotFoundError(
            f"Missing mask directory:\n{mask_class_dir}"
        )

    image_count = len([
        f for f in os.listdir(image_class_dir)
        if f.lower().endswith(
            (".jpg", ".jpeg", ".png")
        )
    ])

    mask_count = len([
        f for f in os.listdir(mask_class_dir)
        if f.lower().endswith(
            (".jpg", ".jpeg", ".png")
        )
    ])

    total_images += image_count
    total_masks += mask_count

    print(
        f"{cls:<18}"
        f"Images: {image_count:<8}"
        f"Masks: {mask_count}"
    )

print("-" * 80)
print(f"Total images : {total_images}")
print(f"Total masks  : {total_masks}")


# ============================================================
# 7. REPRODUCIBILITY
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Deterministic behavior
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

device = torch.device(
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

print("\nUsing device:", device)

if torch.cuda.is_available():

    print(
        "GPU:",
        torch.cuda.get_device_name(0)
    )


# ============================================================
# 8. EXPERIMENT CONFIGURATION
# ============================================================

IMG_SIZE = 256

BATCH_SIZE = 16

# ------------------------------------------------------------
# Pilot setting
# ------------------------------------------------------------
# Use 5 initially to verify the pipeline.
# For final experiment consider 15-30+.
# ------------------------------------------------------------

EPOCHS = 5

LEARNING_RATE = 1e-4

CLS_WEIGHT = 0.5

SEG_WEIGHT = 0.5

NUM_WORKERS = 2

PIN_MEMORY = torch.cuda.is_available()

IMAGENET_MEAN = [
    0.485,
    0.456,
    0.406
]

IMAGENET_STD = [
    0.229,
    0.224,
    0.225
]


# ============================================================
# 9. DATASET CLASS
# ============================================================

class RiceMTLDataset(Dataset):

    def __init__(
        self,
        image_dir,
        mask_dir,
        augment=False
    ):

        self.image_dir = image_dir
        self.mask_dir = mask_dir
        self.augment = augment

        self.image_paths = []
        self.mask_paths = []
        self.labels = []

        self._build_pairs()

    # --------------------------------------------------------
    # Build pairs
    # --------------------------------------------------------

    def _build_pairs(self):

        mask_dict = {}

        for class_name in CLASS_NAMES:

            class_dir = os.path.join(
                self.mask_dir,
                class_name
            )

            if not os.path.isdir(class_dir):
                continue

            for filename in sorted(
                os.listdir(class_dir)
            ):

                if not filename.lower().endswith(
                    (".jpg", ".jpeg", ".png")
                ):
                    continue

                basename = os.path.splitext(
                    filename.lower()
                )[0]

                key = (
                    class_name.lower(),
                    basename
                )

                mask_dict[key] = os.path.join(
                    class_dir,
                    filename
                )

        unmatched_images = []

        for class_name in CLASS_NAMES:

            class_dir = os.path.join(
                self.image_dir,
                class_name
            )

            class_key = class_name.lower()

            label = CLASS_MAP[class_key]

            for filename in sorted(
                os.listdir(class_dir)
            ):

                if not filename.lower().endswith(
                    (".jpg", ".jpeg", ".png")
                ):
                    continue

                basename = os.path.splitext(
                    filename.lower()
                )[0]

                key = (
                    class_key,
                    basename
                )

                image_path = os.path.join(
                    class_dir,
                    filename
                )

                if key not in mask_dict:

                    unmatched_images.append(
                        image_path
                    )

                    continue

                mask_path = mask_dict[key]

                self.image_paths.append(
                    image_path
                )

                self.mask_paths.append(
                    mask_path
                )

                self.labels.append(
                    label
                )

        print(
            f"\n✅ Paired samples: "
            f"{len(self.image_paths)}"
        )

        if unmatched_images:

            print(
                f"⚠️ Images without masks: "
                f"{len(unmatched_images)}"
            )

        counts = np.bincount(
            self.labels,
            minlength=NUM_CLASSES
        )

        print("\nClass distribution:")

        for name, count in zip(
            CLASS_NAMES,
            counts
        ):

            print(
                f"  {name:<18}: {count}"
            )

    # --------------------------------------------------------
    # Length
    # --------------------------------------------------------

    def __len__(self):

        return len(
            self.image_paths
        )

    # --------------------------------------------------------
    # Get sample
    # --------------------------------------------------------

    def __getitem__(self, idx):

        image = Image.open(
            self.image_paths[idx]
        ).convert("RGB")

        mask = Image.open(
            self.mask_paths[idx]
        ).convert("L")

        # ----------------------------------------------------
        # Resize
        # ----------------------------------------------------

        image = TF.resize(
            image,
            [IMG_SIZE, IMG_SIZE],
            interpolation=T.InterpolationMode.BILINEAR
        )

        mask = TF.resize(
            mask,
            [IMG_SIZE, IMG_SIZE],
            interpolation=T.InterpolationMode.NEAREST
        )

        # ----------------------------------------------------
        # Synchronized augmentation
        # ----------------------------------------------------

        if self.augment:

            if random.random() < 0.5:

                image = TF.hflip(image)
                mask = TF.hflip(mask)

            if random.random() < 0.5:

                image = TF.vflip(image)
                mask = TF.vflip(mask)

        # ----------------------------------------------------
        # Image
        # ----------------------------------------------------

        image = TF.to_tensor(image)

        image = TF.normalize(
            image,
            IMAGENET_MEAN,
            IMAGENET_STD
        )

        # ----------------------------------------------------
        # Lesion mask
        # ----------------------------------------------------

        mask = TF.to_tensor(mask)

        # Convert lesion mask to binary
        mask = (
            mask > 0.5
        ).float()

        # ----------------------------------------------------
        # Classification label
        # ----------------------------------------------------

        label = torch.tensor(
            self.labels[idx],
            dtype=torch.long
        )

        return (
            image,
            mask,
            label
        )


# ============================================================
# 10. CREATE FULL DATASET
# ============================================================

full_dataset = RiceMTLDataset(
    image_dir=IMAGE_DIR,
    mask_dir=MASK_DIR,
    augment=False
)

print(
    "\nTotal paired samples:",
    len(full_dataset)
)

assert len(full_dataset) == 5932, (
    f"Expected 5932 samples, "
    f"found {len(full_dataset)}"
)


# ============================================================
# 11. CREATE SAMPLE METADATA
# ============================================================

sample_rows = []

for idx in range(
    len(full_dataset)
):

    sample_rows.append({

        "index": idx,

        "image_path":
            full_dataset.image_paths[idx],

        "mask_path":
            full_dataset.mask_paths[idx],

        "class_id":
            full_dataset.labels[idx],

        "class_name":
            CLASS_NAMES[
                full_dataset.labels[idx]
            ],

        "filename":
            os.path.basename(
                full_dataset.image_paths[idx]
            )

    })


sample_df = pd.DataFrame(
    sample_rows
)

sample_metadata_path = os.path.join(
    RESULTS_DIR,
    "sample_metadata.csv"
)

sample_df.to_csv(
    sample_metadata_path,
    index=False
)

print(
    "\n✅ Sample metadata saved:"
)

print(
    sample_metadata_path
)


# ============================================================
# 12. STRATIFIED TRAIN / VAL / TEST SPLIT
# ============================================================

all_indices = np.arange(
    len(full_dataset)
)

all_labels = np.array(
    full_dataset.labels
)

# ------------------------------------------------------------
# First:
# 70% train
# 30% temporary
# ------------------------------------------------------------

train_idx, temp_idx = train_test_split(
    all_indices,
    test_size=0.30,
    random_state=SEED,
    stratify=all_labels
)

# ------------------------------------------------------------
# Then:
# 15% validation
# 15% test
#
# temp = 30%
# half temp = 15%
# ------------------------------------------------------------

temp_labels = all_labels[temp_idx]

val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.50,
    random_state=SEED,
    stratify=temp_labels
)

print("\n" + "=" * 80)
print("DATA SPLIT")
print("=" * 80)

print(
    f"Train      : {len(train_idx)} "
    f"({len(train_idx)/len(all_indices)*100:.2f}%)"
)

print(
    f"Validation : {len(val_idx)} "
    f"({len(val_idx)/len(all_indices)*100:.2f}%)"
)

print(
    f"Test       : {len(test_idx)} "
    f"({len(test_idx)/len(all_indices)*100:.2f}%)"
)

assert (
    len(set(train_idx) &
        set(val_idx)) == 0
)

assert (
    len(set(train_idx) &
        set(test_idx)) == 0
)

assert (
    len(set(val_idx) &
        set(test_idx)) == 0
)


# ============================================================
# 13. SAVE SPLITS
# ============================================================

split_df = sample_df.copy()

split_df["split"] = "unused"

split_df.loc[
    train_idx,
    "split"
] = "train"

split_df.loc[
    val_idx,
    "split"
] = "validation"

split_df.loc[
    test_idx,
    "split"
] = "test"

split_path = os.path.join(
    RESULTS_DIR,
    "dataset_splits.csv"
)

split_df.to_csv(
    split_path,
    index=False
)

print(
    "\n✅ Split information saved:"
)

print(
    split_path
)


# ============================================================
# 14. CREATE AUGMENTED + VALIDATION DATASETS
# ============================================================

train_dataset = copy.copy(
    full_dataset
)

train_dataset.augment = True

val_dataset = copy.copy(
    full_dataset
)

val_dataset.augment = False

test_dataset = copy.copy(
    full_dataset
)

test_dataset.augment = False


# ============================================================
# 15. DATALOADER FACTORY
# ============================================================

def make_loader(
    dataset,
    indices,
    shuffle=False
):

    subset = Subset(
        dataset,
        indices
    )

    loader = DataLoader(
        subset,
        batch_size=BATCH_SIZE,
        shuffle=shuffle,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY
    )

    return loader


# Full validation/test loaders
val_loader = make_loader(
    val_dataset,
    val_idx,
    shuffle=False
)

test_loader = make_loader(
    test_dataset,
    test_idx,
    shuffle=False
)

print("\nValidation batches:", len(val_loader))
print("Test batches      :", len(test_loader))


# ============================================================
# 16. MODEL 1 — CLASSIFICATION ONLY
# ============================================================

class RiceClsOnlyModel(nn.Module):

    def __init__(
        self,
        num_classes=NUM_CLASSES
    ):

        super().__init__()

        base = smp.Unet(
            encoder_name="resnet34",
            encoder_weights="imagenet",
            in_channels=3,
            classes=1
        )

        self.encoder = base.encoder

        self.pool = nn.AdaptiveAvgPool2d(
            1
        )

        self.fc = nn.Linear(
            512,
            num_classes
        )

    def forward(self, x):

        features = self.encoder(x)

        deepest = features[-1]

        pooled = self.pool(
            deepest
        ).flatten(1)

        return self.fc(
            pooled
        )


# ============================================================
# 17. MODEL 2 — SEGMENTATION ONLY
# ============================================================

class RiceSegOnlyModel(nn.Module):

    def __init__(self):

        super().__init__()

        self.unet = smp.Unet(
            encoder_name="resnet34",
            encoder_weights="imagenet",
            in_channels=3,
            classes=1
        )

    def forward(self, x):

        return self.unet(x)


# ============================================================
# 18. MODEL 3 — MULTI-TASK
# ============================================================

def get_mtl_model(
    num_classes=NUM_CLASSES
):

    return smp.Unet(

        encoder_name="resnet34",

        encoder_weights="imagenet",

        in_channels=3,

        classes=1,

        aux_params={

            "pooling": "avg",

            "dropout": 0.3,

            "activation": None,

            "classes": num_classes

        }
    )


# ============================================================
# 19. LOSS FUNCTIONS
# ============================================================

criterion_cls = nn.CrossEntropyLoss()

criterion_seg = smp.losses.DiceLoss(
    mode=smp.losses.BINARY_MODE,
    from_logits=True
)


# ============================================================
# 20. METRICS
# ============================================================

def compute_dice_iou(
    pred_logits,
    target,
    eps=1e-7
):

    pred = (
        torch.sigmoid(
            pred_logits
        ) > 0.5
    ).float()

    target = target.float()

    intersection = (
        pred * target
    ).sum(
        dim=(1, 2, 3)
    )

    pred_area = pred.sum(
        dim=(1, 2, 3)
    )

    target_area = target.sum(
        dim=(1, 2, 3)
    )

    dice = (
        2.0 * intersection + eps
    ) / (
        pred_area +
        target_area +
        eps
    )

    union = (
        pred_area +
        target_area -
        intersection
    )

    iou = (
        intersection + eps
    ) / (
        union + eps
    )

    return (
        dice.mean().item(),
        iou.mean().item()
    )


# ============================================================
# 21. PARTIAL-LABEL TASK ASSIGNMENT
# ============================================================

def create_partial_task_assignment(
    train_indices,
    ratio_cls,
    seed
):
    """
    Creates disjoint task supervision.

    ratio_cls = fraction receiving CLS supervision.
    The remaining fraction receives SEG supervision.

    Example:
        0.50 -> 50% CLS / 50% SEG
        0.70 -> 70% CLS / 30% SEG
        0.30 -> 30% CLS / 70% SEG

    Sampling is stratified by disease class.
    """

    train_indices = np.array(
        train_indices
    )

    train_labels = all_labels[
        train_indices
    ]

    cls_indices, seg_indices = train_test_split(

        train_indices,

        test_size=(1.0 - ratio_cls),

        random_state=seed,

        stratify=train_labels
    )

    cls_indices = np.sort(
        cls_indices
    )

    seg_indices = np.sort(
        seg_indices
    )

    # Verify disjointness
    assert (
        len(
            set(cls_indices) &
            set(seg_indices)
        ) == 0
    )

    # Verify complete coverage
    assert (
        set(cls_indices) |
        set(seg_indices)
    ) == set(train_indices)

    assignment = {}

    for idx in cls_indices:

        assignment[
            int(idx)
        ] = {
            "classification": 1,
            "segmentation": 0
        }

    for idx in seg_indices:

        assignment[
            int(idx)
        ] = {
            "classification": 0,
            "segmentation": 1
        }

    return (
        assignment,
        cls_indices,
        seg_indices
    )


# ============================================================
# 22. SAVE TASK ASSIGNMENT
# ============================================================

def save_task_assignment(
    assignment,
    cls_indices,
    seg_indices,
    ratio_name
):

    df = sample_df[
        sample_df["index"].isin(
            train_idx
        )
    ].copy()

    df["classification_available"] = (
        df["index"].map(
            lambda x:
            assignment[
                int(x)
            ]["classification"]
        )
    )

    df["segmentation_available"] = (
        df["index"].map(
            lambda x:
            assignment[
                int(x)
            ]["segmentation"]
        )
    )

    df["supervision_group"] = (
        df.apply(
            lambda row:
            "classification_only"
            if row["classification_available"] == 1
            else "segmentation_only",
            axis=1
        )
    )

    path = os.path.join(
        RESULTS_DIR,
        f"task_assignment_{ratio_name}.csv"
    )

    df.to_csv(
        path,
        index=False
    )

    return path


# ============================================================
# 23. GENERIC BATCH RUNNER
# ============================================================

def run_epoch_stl_cls(
    model,
    loader,
    optimizer=None
):

    is_train = (
        optimizer is not None
    )

    if is_train:
        model.train()
    else:
        model.eval()

    total_loss = 0.0

    all_preds = []
    all_labels = []

    num_batches = 0

    context = (
        torch.enable_grad()
        if is_train
        else torch.no_grad()
    )

    with context:

        for images, _, labels in loader:

            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            if is_train:

                optimizer.zero_grad(
                    set_to_none=True
                )

            logits = model(
                images
            )

            loss = criterion_cls(
                logits,
                labels
            )

            if is_train:

                loss.backward()

                optimizer.step()

            preds = logits.argmax(
                dim=1
            )

            all_preds.extend(
                preds.cpu().tolist()
            )

            all_labels.extend(
                labels.cpu().tolist()
            )

            total_loss += loss.item()

            num_batches += 1

    metrics = {

        "loss":
            total_loss /
            max(num_batches, 1),

        "acc":
            accuracy_score(
                all_labels,
                all_preds
            ),

        "f1_macro":
            f1_score(
                all_labels,
                all_preds,
                average="macro"
            )
    }

    return metrics


# ============================================================
# 24. STL SEGMENTATION EPOCH
# ============================================================

def run_epoch_stl_seg(
    model,
    loader,
    optimizer=None
):

    is_train = (
        optimizer is not None
    )

    if is_train:
        model.train()
    else:
        model.eval()

    total_loss = 0.0

    dice_scores = []
    iou_scores = []

    num_batches = 0

    context = (
        torch.enable_grad()
        if is_train
        else torch.no_grad()
    )

    with context:

        for images, masks, _ in loader:

            images = images.to(
                device,
                non_blocking=True
            )

            masks = masks.to(
                device,
                non_blocking=True
            )

            if is_train:

                optimizer.zero_grad(
                    set_to_none=True
                )

            logits = model(
                images
            )

            loss = criterion_seg(
                logits,
                masks
            )

            if is_train:

                loss.backward()

                optimizer.step()

            dice, iou = compute_dice_iou(
                logits,
                masks
            )

            dice_scores.append(
                dice
            )

            iou_scores.append(
                iou
            )

            total_loss += loss.item()

            num_batches += 1

    metrics = {

        "loss":
            total_loss /
            max(num_batches, 1),

        "dice":
            float(
                np.mean(dice_scores)
            ),

        "iou":
            float(
                np.mean(iou_scores)
            )
    }

    return metrics


# ============================================================
# 25. FULL MTL EPOCH
# ============================================================

def run_epoch_mtl(
    model,
    loader,
    optimizer=None
):

    is_train = (
        optimizer is not None
    )

    if is_train:
        model.train()
    else:
        model.eval()

    total_loss = 0.0

    total_cls_loss = 0.0

    total_seg_loss = 0.0

    all_preds = []

    all_labels = []

    dice_scores = []

    iou_scores = []

    num_batches = 0

    context = (
        torch.enable_grad()
        if is_train
        else torch.no_grad()
    )

    with context:

        for images, masks, labels in loader:

            images = images.to(
                device,
                non_blocking=True
            )

            masks = masks.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            if is_train:

                optimizer.zero_grad(
                    set_to_none=True
                )

            seg_out, cls_out = model(
                images
            )

            loss_cls = criterion_cls(
                cls_out,
                labels
            )

            loss_seg = criterion_seg(
                seg_out,
                masks
            )

            loss = (
                CLS_WEIGHT * loss_cls
                +
                SEG_WEIGHT * loss_seg
            )

            if is_train:

                loss.backward()

                optimizer.step()

            predictions = cls_out.argmax(
                dim=1
            )

            all_preds.extend(
                predictions.cpu().tolist()
            )

            all_labels.extend(
                labels.cpu().tolist()
            )

            dice, iou = compute_dice_iou(
                seg_out,
                masks
            )

            dice_scores.append(
                dice
            )

            iou_scores.append(
                iou
            )

            total_loss += loss.item()

            total_cls_loss += loss_cls.item()

            total_seg_loss += loss_seg.item()

            num_batches += 1

    metrics = {

        "loss":
            total_loss /
            max(num_batches, 1),

        "cls_loss":
            total_cls_loss /
            max(num_batches, 1),

        "seg_loss":
            total_seg_loss /
            max(num_batches, 1),

        "acc":
            accuracy_score(
                all_labels,
                all_preds
            ),

        "f1_macro":
            f1_score(
                all_labels,
                all_preds,
                average="macro"
            ),

        "dice":
            float(
                np.mean(dice_scores)
            ),

        "iou":
            float(
                np.mean(iou_scores)
            )
    }

    return metrics


# ============================================================
# 26. PARTIAL-LABEL MTL EPOCH
# ============================================================

def run_epoch_partial_mtl(
    model,
    loader,
    task_assignment,
    optimizer=None
):
    """
    Each sample has:
        classification_available = 0/1
        segmentation_available   = 0/1

    The two labels are disjoint in our proposed experiments.

    Crucial:
        Loss is normalized by ACTIVE task weights.

    Therefore a batch containing only CLS labels does not
    accidentally receive half-strength classification gradients.
    """

    is_train = (
        optimizer is not None
    )

    if is_train:
        model.train()
    else:
        model.eval()

    total_loss = 0.0

    total_cls_loss = 0.0

    total_seg_loss = 0.0

    cls_batches = 0
    seg_batches = 0

    cls_preds = []
    cls_labels = []

    dice_scores = []
    iou_scores = []

    num_batches = 0

    context = (
        torch.enable_grad()
        if is_train
        else torch.no_grad()
    )

    with context:

        for batch in loader:

            images, masks, labels, indices = (
                batch
            )

            images = images.to(
                device,
                non_blocking=True
            )

            masks = masks.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            batch_indices = (
                indices.numpy()
            )

            cls_available = torch.tensor(
                [
                    task_assignment[
                        int(idx)
                    ]["classification"]
                    for idx in batch_indices
                ],
                dtype=torch.bool,
                device=device
            )

            seg_available = torch.tensor(
                [
                    task_assignment[
                        int(idx)
                    ]["segmentation"]
                    for idx in batch_indices
                ],
                dtype=torch.bool,
                device=device
            )

            if is_train:

                optimizer.zero_grad(
                    set_to_none=True
                )

            seg_out, cls_out = model(
                images
            )

            active_losses = []

            # ------------------------------------------------
            # Classification loss
            # ------------------------------------------------

            if cls_available.any():

                cls_logits_active = (
                    cls_out[
                        cls_available
                    ]
                )

                labels_active = (
                    labels[
                        cls_available
                    ]
                )

                loss_cls = criterion_cls(
                    cls_logits_active,
                    labels_active
                )

                active_losses.append(
                    CLS_WEIGHT * loss_cls
                )

                total_cls_loss += (
                    loss_cls.item()
                )

                cls_batches += 1

                cls_predictions = (
                    cls_logits_active.argmax(
                        dim=1
                    )
                )

                cls_preds.extend(
                    cls_predictions
                    .detach()
                    .cpu()
                    .tolist()
                )

                cls_labels.extend(
                    labels_active
                    .detach()
                    .cpu()
                    .tolist()
                )

            # ------------------------------------------------
            # Segmentation loss
            # ------------------------------------------------

            if seg_available.any():

                seg_logits_active = (
                    seg_out[
                        seg_available
                    ]
                )

                masks_active = (
                    masks[
                        seg_available
                    ]
                )

                loss_seg = criterion_seg(
                    seg_logits_active,
                    masks_active
                )

                active_losses.append(
                    SEG_WEIGHT * loss_seg
                )

                total_seg_loss += (
                    loss_seg.item()
                )

                seg_batches += 1

                dice, iou = (
                    compute_dice_iou(
                        seg_logits_active,
                        masks_active
                    )
                )

                dice_scores.append(
                    dice
                )

                iou_scores.append(
                    iou
                )

            # ------------------------------------------------
            # Normalize by active weights
            # ------------------------------------------------

            if len(active_losses) == 0:

                continue

            active_weight_sum = 0.0

            if cls_available.any():

                active_weight_sum += (
                    CLS_WEIGHT
                )

            if seg_available.any():

                active_weight_sum += (
                    SEG_WEIGHT
                )

            loss = (
                torch.stack(
                    active_losses
                ).sum()
                /
                active_weight_sum
            )

            if is_train:

                loss.backward()

                optimizer.step()

            total_loss += loss.item()

            num_batches += 1

    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    metrics = {

        "loss":
            total_loss /
            max(num_batches, 1)
    }

    if cls_labels:

        metrics["acc"] = (
            accuracy_score(
                cls_labels,
                cls_preds
            )
        )

        metrics["f1_macro"] = (
            f1_score(
                cls_labels,
                cls_preds,
                average="macro"
            )
        )

    if dice_scores:

        metrics["dice"] = float(
            np.mean(dice_scores)
        )

        metrics["iou"] = float(
            np.mean(iou_scores)
        )

    metrics["cls_loss"] = (
        total_cls_loss /
        max(cls_batches, 1)
    )

    metrics["seg_loss"] = (
        total_seg_loss /
        max(seg_batches, 1)
    )

    return metrics


# ============================================================
# 27. INDEX-AWARE DATASET
# ============================================================

class IndexDataset(Dataset):

    """
    Wraps RiceMTLDataset and returns original dataset index.
    Required for task supervision lookup in Partial MTL.
    """

    def __init__(
        self,
        base_dataset
    ):

        self.base_dataset = base_dataset

    def __len__(self):

        return len(
            self.base_dataset
        )

    def __getitem__(
        self,
        idx
    ):

        image, mask, label = (
            self.base_dataset[idx]
        )

        return (
            image,
            mask,
            label,
            idx
        )


# Create indexed training dataset
indexed_train_dataset = IndexDataset(
    train_dataset
)


def make_partial_loader(
    indices,
    shuffle=True
):

    subset = Subset(
        indexed_train_dataset,
        indices
    )

    return DataLoader(
        subset,
        batch_size=BATCH_SIZE,
        shuffle=shuffle,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY
    )


# ============================================================
# 28. GENERIC TRAINING — CLASSIFICATION
# ============================================================

def train_cls_model(
    train_loader,
    val_loader,
    epochs=EPOCHS,
    lr=LEARNING_RATE,
    experiment_name="cls"
):

    model = RiceClsOnlyModel(
        num_classes=NUM_CLASSES
    ).to(device)

    optimizer = optim.Adam(
        model.parameters(),
        lr=lr
    )

    best_val_loss = float("inf")

    best_state = None

    best_epoch = None

    history = []

    for epoch in range(
        1,
        epochs + 1
    ):

        start_time = time.time()

        train_metrics = (
            run_epoch_stl_cls(
                model,
                train_loader,
                optimizer
            )
        )

        val_metrics = (
            run_epoch_stl_cls(
                model,
                val_loader,
                optimizer=None
            )
        )

        elapsed = (
            time.time() -
            start_time
        )

        history.append({
            "epoch": epoch,
            "train": train_metrics,
            "val": val_metrics
        })

        print(
            f"[{experiment_name}] "
            f"Epoch {epoch}/{epochs} | "
            f"Train Loss {train_metrics['loss']:.4f} | "
            f"Val Loss {val_metrics['loss']:.4f} | "
            f"Val Acc {val_metrics['acc']*100:.2f}% | "
            f"Val F1 {val_metrics['f1_macro']:.4f} | "
            f"{elapsed:.1f}s"
        )

        if (
            val_metrics["loss"]
            <
            best_val_loss
        ):

            best_val_loss = (
                val_metrics["loss"]
            )

            best_state = copy.deepcopy(
                model.state_dict()
            )

            best_epoch = epoch

    if best_state is not None:

        model.load_state_dict(
            best_state
        )

    # Save checkpoint
    checkpoint_path = os.path.join(
        SAVE_DIR,
        f"{experiment_name}_best.pth"
    )

    torch.save(
        model.state_dict(),
        checkpoint_path
    )

    print(
        f"✅ Best epoch: {best_epoch}"
    )

    print(
        f"✅ Saved: {checkpoint_path}"
    )

    return (
        model,
        history,
        best_epoch,
        checkpoint_path
    )


# ============================================================
# 29. GENERIC TRAINING — SEGMENTATION
# ============================================================

def train_seg_model(
    train_loader,
    val_loader,
    epochs=EPOCHS,
    lr=LEARNING_RATE,
    experiment_name="seg"
):

    model = RiceSegOnlyModel().to(
        device
    )

    optimizer = optim.Adam(
        model.parameters(),
        lr=lr
    )

    best_val_loss = float("inf")

    best_state = None

    best_epoch = None

    history = []

    for epoch in range(
        1,
        epochs + 1
    ):

        start_time = time.time()

        train_metrics = (
            run_epoch_stl_seg(
                model,
                train_loader,
                optimizer
            )
        )

        val_metrics = (
            run_epoch_stl_seg(
                model,
                val_loader,
                optimizer=None
            )
        )

        elapsed = (
            time.time() -
            start_time
        )

        history.append({
            "epoch": epoch,
            "train": train_metrics,
            "val": val_metrics
        })

        print(
            f"[{experiment_name}] "
            f"Epoch {epoch}/{epochs} | "
            f"Train Loss {train_metrics['loss']:.4f} | "
            f"Val Loss {val_metrics['loss']:.4f} | "
            f"Val Dice {val_metrics['dice']:.4f} | "
            f"Val IoU {val_metrics['iou']:.4f} | "
            f"{elapsed:.1f}s"
        )

        if (
            val_metrics["loss"]
            <
            best_val_loss
        ):

            best_val_loss = (
                val_metrics["loss"]
            )

            best_state = copy.deepcopy(
                model.state_dict()
            )

            best_epoch = epoch

    if best_state is not None:

        model.load_state_dict(
            best_state
        )

    checkpoint_path = os.path.join(
        SAVE_DIR,
        f"{experiment_name}_best.pth"
    )

    torch.save(
        model.state_dict(),
        checkpoint_path
    )

    print(
        f"✅ Best epoch: {best_epoch}"
    )

    print(
        f"✅ Saved: {checkpoint_path}"
    )

    return (
        model,
        history,
        best_epoch,
        checkpoint_path
    )


# ============================================================
# 30. GENERIC TRAINING — FULL MTL
# ============================================================

def train_full_mtl_model(
    train_loader,
    val_loader,
    epochs=EPOCHS,
    lr=LEARNING_RATE,
    experiment_name="full_mtl"
):

    model = get_mtl_model(
        num_classes=NUM_CLASSES
    ).to(device)

    optimizer = optim.Adam(
        model.parameters(),
        lr=lr
    )

    best_val_loss = float("inf")

    best_state = None

    best_epoch = None

    history = []

    for epoch in range(
        1,
        epochs + 1
    ):

        start_time = time.time()

        train_metrics = (
            run_epoch_mtl(
                model,
                train_loader,
                optimizer
            )
        )

        val_metrics = (
            run_epoch_mtl(
                model,
                val_loader,
                optimizer=None
            )
        )

        elapsed = (
            time.time() -
            start_time
        )

        history.append({
            "epoch": epoch,
            "train": train_metrics,
            "val": val_metrics
        })

        print(
            f"[{experiment_name}] "
            f"Epoch {epoch}/{epochs} | "
            f"Train Loss {train_metrics['loss']:.4f} | "
            f"Val Loss {val_metrics['loss']:.4f} | "
            f"Acc {val_metrics['acc']*100:.2f}% | "
            f"F1 {val_metrics['f1_macro']:.4f} | "
            f"Dice {val_metrics['dice']:.4f} | "
            f"IoU {val_metrics['iou']:.4f} | "
            f"{elapsed:.1f}s"
        )

        if (
            val_metrics["loss"]
            <
            best_val_loss
        ):

            best_val_loss = (
                val_metrics["loss"]
            )

            best_state = copy.deepcopy(
                model.state_dict()
            )

            best_epoch = epoch

    if best_state is not None:

        model.load_state_dict(
            best_state
        )

    checkpoint_path = os.path.join(
        SAVE_DIR,
        f"{experiment_name}_best.pth"
    )

    torch.save(
        model.state_dict(),
        checkpoint_path
    )

    print(
        f"✅ Best epoch: {best_epoch}"
    )

    print(
        f"✅ Saved: {checkpoint_path}"
    )

    return (
        model,
        history,
        best_epoch,
        checkpoint_path
    )


# ============================================================
# 31. GENERIC TRAINING — PARTIAL MTL
# ============================================================

def train_partial_mtl_model(
    train_loader,
    val_loader,
    task_assignment,
    epochs=EPOCHS,
    lr=LEARNING_RATE,
    experiment_name="partial_mtl"
):

    model = get_mtl_model(
        num_classes=NUM_CLASSES
    ).to(device)

    optimizer = optim.Adam(
        model.parameters(),
        lr=lr
    )

    best_val_loss = float("inf")

    best_state = None

    best_epoch = None

    history = []

    for epoch in range(
        1,
        epochs + 1
    ):

        start_time = time.time()

        train_metrics = (
            run_epoch_partial_mtl(
                model,
                train_loader,
                task_assignment,
                optimizer
            )
        )

        # ----------------------------------------------------
        # IMPORTANT:
        # Validation is FULL-LABEL.
        # No labels are hidden in validation.
        # ----------------------------------------------------

        val_metrics = (
            run_epoch_mtl(
                model,
                val_loader,
                optimizer=None
            )
        )

        elapsed = (
            time.time() -
            start_time
        )

        history.append({
            "epoch": epoch,
            "train": train_metrics,
            "val": val_metrics
        })

        print(
            f"[{experiment_name}] "
            f"Epoch {epoch}/{epochs} | "
            f"Train Loss {train_metrics['loss']:.4f} | "
            f"Val Loss {val_metrics['loss']:.4f} | "
            f"Val Acc {val_metrics['acc']*100:.2f}% | "
            f"Val F1 {val_metrics['f1_macro']:.4f} | "
            f"Val Dice {val_metrics['dice']:.4f} | "
            f"Val IoU {val_metrics['iou']:.4f} | "
            f"{elapsed:.1f}s"
        )

        if (
            val_metrics["loss"]
            <
            best_val_loss
        ):

            best_val_loss = (
                val_metrics["loss"]
            )

            best_state = copy.deepcopy(
                model.state_dict()
            )

            best_epoch = epoch

    if best_state is not None:

        model.load_state_dict(
            best_state
        )

    checkpoint_path = os.path.join(
        SAVE_DIR,
        f"{experiment_name}_best.pth"
    )

    torch.save(
        model.state_dict(),
        checkpoint_path
    )

    print(
        f"✅ Best epoch: {best_epoch}"
    )

    print(
        f"✅ Saved: {checkpoint_path}"
    )

    return (
        model,
        history,
        best_epoch,
        checkpoint_path
    )


# ============================================================
# 32. TEST EVALUATION
# ============================================================

def evaluate_cls_test(
    model,
    test_loader
):

    model.eval()

    all_preds = []

    all_labels = []

    losses = []

    with torch.no_grad():

        for images, _, labels in test_loader:

            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            logits = model(
                images
            )

            loss = criterion_cls(
                logits,
                labels
            )

            losses.append(
                loss.item()
            )

            preds = logits.argmax(
                dim=1
            )

            all_preds.extend(
                preds.cpu().tolist()
            )

            all_labels.extend(
                labels.cpu().tolist()
            )

    return {

        "loss":
            float(np.mean(losses)),

        "accuracy":
            accuracy_score(
                all_labels,
                all_preds
            ),

        "f1_macro":
            f1_score(
                all_labels,
                all_preds,
                average="macro"
            )
    }


# ------------------------------------------------------------
# Segmentation test
# ------------------------------------------------------------

def evaluate_seg_test(
    model,
    test_loader
):

    model.eval()

    losses = []

    dices = []

    ious = []

    with torch.no_grad():

        for images, masks, _ in test_loader:

            images = images.to(
                device,
                non_blocking=True
            )

            masks = masks.to(
                device,
                non_blocking=True
            )

            logits = model(
                images
            )

            loss = criterion_seg(
                logits,
                masks
            )

            dice, iou = (
                compute_dice_iou(
                    logits,
                    masks
                )
            )

            losses.append(
                loss.item()
            )

            dices.append(dice)

            ious.append(iou)

    return {

        "loss":
            float(np.mean(losses)),

        "dice":
            float(np.mean(dices)),

        "iou":
            float(np.mean(ious))
    }


# ------------------------------------------------------------
# Full MTL test
# ------------------------------------------------------------

def evaluate_mtl_test(
    model,
    test_loader
):

    model.eval()

    total_losses = []

    cls_losses = []

    seg_losses = []

    preds = []

    labels_all = []

    dices = []

    ious = []

    with torch.no_grad():

        for images, masks, labels in test_loader:

            images = images.to(
                device,
                non_blocking=True
            )

            masks = masks.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            seg_out, cls_out = model(
                images
            )

            loss_cls = criterion_cls(
                cls_out,
                labels
            )

            loss_seg = criterion_seg(
                seg_out,
                masks
            )

            total_loss = (
                CLS_WEIGHT * loss_cls
                +
                SEG_WEIGHT * loss_seg
            )

            total_losses.append(
                total_loss.item()
            )

            cls_losses.append(
                loss_cls.item()
            )

            seg_losses.append(
                loss_seg.item()
            )

            batch_preds = (
                cls_out.argmax(
                    dim=1
                )
            )

            preds.extend(
                batch_preds.cpu().tolist()
            )

            labels_all.extend(
                labels.cpu().tolist()
            )

            dice, iou = (
                compute_dice_iou(
                    seg_out,
                    masks
                )
            )

            dices.append(dice)
            ious.append(iou)

    return {

        "loss":
            float(
                np.mean(total_losses)
            ),

        "cls_loss":
            float(
                np.mean(cls_losses)
            ),

        "seg_loss":
            float(
                np.mean(seg_losses)
            ),

        "accuracy":
            accuracy_score(
                labels_all,
                preds
            ),

        "f1_macro":
            f1_score(
                labels_all,
                preds,
                average="macro"
            ),

        "dice":
            float(
                np.mean(dices)
            ),

        "iou":
            float(
                np.mean(ious)
            )
    }


# ============================================================
# 33. TEST LOADER FOR SUBSETS
# ============================================================

def make_subset_loader(
    dataset,
    indices
):

    subset = Subset(
        dataset,
        np.array(indices)
    )

    return DataLoader(
        subset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=PIN_MEMORY
    )


# ============================================================
# 34. RESULTS STORAGE
# ============================================================

all_results = []

all_histories = {}

all_models = {}


def store_history(
    experiment_name,
    history
):

    all_histories[
        experiment_name
    ] = history

    history_path = os.path.join(
        RESULTS_DIR,
        f"{experiment_name}_history.json"
    )

    with open(
        history_path,
        "w"
    ) as f:

        json.dump(
            history,
            f,
            indent=2
        )


def add_result(
    experiment_name,
    experiment_type,
    supervision,
    best_epoch,
    test_metrics
):

    row = {

        "Experiment":
            experiment_name,

        "Type":
            experiment_type,

        "Classification_supervision":
            supervision.get(
                "classification",
                np.nan
            ),

        "Segmentation_supervision":
            supervision.get(
                "segmentation",
                np.nan
            ),

        "Best_epoch":
            best_epoch,

        "Test_Loss":
            test_metrics.get(
                "loss",
                np.nan
            ),

        "Test_Accuracy":
            test_metrics.get(
                "accuracy",
                np.nan
            ),

        "Test_F1_Macro":
            test_metrics.get(
                "f1_macro",
                np.nan
            ),

        "Test_Dice":
            test_metrics.get(
                "dice",
                np.nan
            ),

        "Test_IoU":
            test_metrics.get(
                "iou",
                np.nan
            )
    }

    all_results.append(
        row
    )


# ============================================================
# 35. EXPERIMENT E1 — STL CLASSIFICATION 100%
# ============================================================

print("\n\n")
print("=" * 80)
print("E1 — STL CLASSIFICATION — 100% TRAIN LABELS")
print("=" * 80)

full_cls_train_loader = make_loader(
    train_dataset,
    train_idx,
    shuffle=True
)

cls_model, cls_history, cls_best_epoch, cls_checkpoint = (
    train_cls_model(

        train_loader=
            full_cls_train_loader,

        val_loader=
            val_loader,

        experiment_name=
            "E1_STL_CLS_100"

    )
)

cls_test = evaluate_cls_test(
    cls_model,
    test_loader
)

print("\nTEST RESULT")
print(cls_test)

store_history(
    "E1_STL_CLS_100",
    cls_history
)

add_result(
    "E1_STL_CLS_100",
    "STL",
    {
        "classification": 1.0,
        "segmentation": np.nan
    },
    cls_best_epoch,
    cls_test
)

all_models[
    "E1_STL_CLS_100"
] = cls_model


# ============================================================
# 36. EXPERIMENT E2 — STL SEGMENTATION 100%
# ============================================================

print("\n\n")
print("=" * 80)
print("E2 — STL SEGMENTATION — 100% TRAIN LABELS")
print("=" * 80)

full_seg_train_loader = make_loader(
    train_dataset,
    train_idx,
    shuffle=True
)

seg_model, seg_history, seg_best_epoch, seg_checkpoint = (
    train_seg_model(

        train_loader=
            full_seg_train_loader,

        val_loader=
            val_loader,

        experiment_name=
            "E2_STL_SEG_100"

    )
)

seg_test = evaluate_seg_test(
    seg_model,
    test_loader
)

print("\nTEST RESULT")
print(seg_test)

store_history(
    "E2_STL_SEG_100",
    seg_history
)

add_result(
    "E2_STL_SEG_100",
    "STL",
    {
        "classification": np.nan,
        "segmentation": 1.0
    },
    seg_best_epoch,
    seg_test
)

all_models[
    "E2_STL_SEG_100"
] = seg_model


# ============================================================
# 37. EXPERIMENT E3 — FULL-LABEL MTL
# ============================================================

print("\n\n")
print("=" * 80)
print("E3 — FULL-LABEL MTL — 100% + 100%")
print("=" * 80)

full_mtl_train_loader = make_loader(
    train_dataset,
    train_idx,
    shuffle=True
)

full_mtl_model, full_mtl_history, full_mtl_best_epoch, full_mtl_checkpoint = (
    train_full_mtl_model(

        train_loader=
            full_mtl_train_loader,

        val_loader=
            val_loader,

        experiment_name=
            "E3_FULL_MTL"

    )
)

full_mtl_test = evaluate_mtl_test(
    full_mtl_model,
    test_loader
)

print("\nTEST RESULT")
print(full_mtl_test)

store_history(
    "E3_FULL_MTL",
    full_mtl_history
)

add_result(
    "E3_FULL_MTL",
    "FULL_MTL",
    {
        "classification": 1.0,
        "segmentation": 1.0
    },
    full_mtl_best_epoch,
    full_mtl_test
)

all_models[
    "E3_FULL_MTL"
] = full_mtl_model


# ============================================================
# 38. FUNCTION FOR PARTIAL-MTL EXPERIMENT
# ============================================================

def run_partial_experiment(
    ratio_cls,
    ratio_name,
    experiment_id,
    run_controls=True
):

    print("\n\n")
    print("=" * 80)
    print(
        f"{experiment_id} — "
        f"PARTIAL MTL — "
        f"{int(ratio_cls*100)}/{int((1-ratio_cls)*100)}"
    )
    print("=" * 80)

    # --------------------------------------------------------
    # Create disjoint task groups
    # --------------------------------------------------------

    assignment, cls_indices, seg_indices = (
        create_partial_task_assignment(

            train_idx,

            ratio_cls=ratio_cls,

            seed=SEED + int(ratio_cls * 100)

        )
    )

    print(
        f"\nClassification-supervised samples: "
        f"{len(cls_indices)}"
    )

    print(
        f"Segmentation-supervised samples: "
        f"{len(seg_indices)}"
    )

    # --------------------------------------------------------
    # Save task assignment
    # --------------------------------------------------------

    assignment_path = (
        save_task_assignment(

            assignment,

            cls_indices,

            seg_indices,

            ratio_name

        )
    )

    print(
        "\nTask assignment saved:"
    )

    print(
        assignment_path
    )

    # --------------------------------------------------------
    # Partial MTL training loader
    # --------------------------------------------------------

    partial_loader = (
        make_partial_loader(
            train_idx,
            shuffle=True
        )
    )

    # --------------------------------------------------------
    # Train partial MTL
    # --------------------------------------------------------

    experiment_name = (
        f"{experiment_id}_PARTIAL_MTL_{ratio_name}"
    )

    model, history, best_epoch, checkpoint = (
        train_partial_mtl_model(

            train_loader=
                partial_loader,

            val_loader=
                val_loader,

            task_assignment=
                assignment,

            experiment_name=
                experiment_name

        )
    )

    # --------------------------------------------------------
    # Test
    # --------------------------------------------------------

    test_metrics = evaluate_mtl_test(
        model,
        test_loader
    )

    print("\nTEST RESULT")
    print(test_metrics)

    store_history(
        experiment_name,
        history
    )

    add_result(
        experiment_name,
        "PARTIAL_MTL",
        {
            "classification":
                ratio_cls,

            "segmentation":
                1.0 - ratio_cls
        },
        best_epoch,
        test_metrics
    )

    all_models[
        experiment_name
    ] = model

    # --------------------------------------------------------
    # Label-matched controls
    # --------------------------------------------------------

    if run_controls:

        # ====================================================
        # Classification control
        # ====================================================

        cls_control_name = (
            f"{experiment_id}_STL_CLS_"
            f"{int(ratio_cls*100)}"
        )

        print("\n")
        print("-" * 80)
        print(
            f"{cls_control_name}"
        )
        print("-" * 80)

        cls_control_loader = make_loader(
            train_dataset,
            cls_indices,
            shuffle=True
        )

        cls_ctrl_model, cls_ctrl_history, cls_ctrl_best_epoch, cls_ctrl_checkpoint = (
            train_cls_model(

                train_loader=
                    cls_control_loader,

                val_loader=
                    val_loader,

                experiment_name=
                    cls_control_name

            )
        )

        cls_ctrl_test = evaluate_cls_test(
            cls_ctrl_model,
            test_loader
        )

        print(
            "\nTEST RESULT"
        )

        print(
            cls_ctrl_test
        )

        store_history(
            cls_control_name,
            cls_ctrl_history
        )

        add_result(
            cls_control_name,
            "LABEL_MATCHED_STL_CLS",
            {
                "classification":
                    ratio_cls,

                "segmentation":
                    np.nan
            },
            cls_ctrl_best_epoch,
            cls_ctrl_test
        )

        all_models[
            cls_control_name
        ] = cls_ctrl_model

        # ====================================================
        # Segmentation control
        # ====================================================

        seg_ratio = (
            1.0 - ratio_cls
        )

        seg_control_name = (
            f"{experiment_id}_STL_SEG_"
            f"{int(seg_ratio*100)}"
        )

        print("\n")
        print("-" * 80)
        print(
            f"{seg_control_name}"
        )
        print("-" * 80)

        seg_control_loader = make_loader(
            train_dataset,
            seg_indices,
            shuffle=True
        )

        seg_ctrl_model, seg_ctrl_history, seg_ctrl_best_epoch, seg_ctrl_checkpoint = (
            train_seg_model(

                train_loader=
                    seg_control_loader,

                val_loader=
                    val_loader,

                experiment_name=
                    seg_control_name

            )
        )

        seg_ctrl_test = evaluate_seg_test(
            seg_ctrl_model,
            test_loader
        )

        print(
            "\nTEST RESULT"
        )

        print(
            seg_ctrl_test
        )

        store_history(
            seg_control_name,
            seg_ctrl_history
        )

        add_result(
            seg_control_name,
            "LABEL_MATCHED_STL_SEG",
            {
                "classification":
                    np.nan,

                "segmentation":
                    seg_ratio
            },
            seg_ctrl_best_epoch,
            seg_ctrl_test
        )

        all_models[
            seg_control_name
        ] = seg_ctrl_model

    return {
        "partial_model": model,
        "assignment": assignment,
        "cls_indices": cls_indices,
        "seg_indices": seg_indices
    }


# ============================================================
# 39. EXPERIMENT E4/E5/E6 — 50/50
# ============================================================

exp_50 = run_partial_experiment(
    ratio_cls=0.50,
    ratio_name="50_50",
    experiment_id="E4",
    run_controls=True
)


# ============================================================
# 40. EXPERIMENT E7/E8/E9 — 70/30
# ============================================================

exp_70 = run_partial_experiment(
    ratio_cls=0.70,
    ratio_name="70_30",
    experiment_id="E7",
    run_controls=True
)


# ============================================================
# 41. EXPERIMENT E10/E11/E12 — 30/70
# ============================================================

exp_30 = run_partial_experiment(
    ratio_cls=0.30,
    ratio_name="30_70",
    experiment_id="E10",
    run_controls=True
)


# ============================================================
# 42. SAVE FINAL RESULTS
# ============================================================

results_df = pd.DataFrame(
    all_results
)

results_path = os.path.join(
    RESULTS_DIR,
    "FINAL_PARTIAL_LABEL_MTL_RESULTS.csv"
)

results_df.to_csv(
    results_path,
    index=False
)

print("\n\n")
print("=" * 120)
print("FINAL EXPERIMENT RESULTS")
print("=" * 120)

display(
    results_df
)

print(
    "\n✅ Final results saved to:"
)

print(
    results_path
)


# ============================================================
# 43. CLEAN COMPARISON TABLES
# ============================================================

# ------------------------------------------------------------
# Classification comparison
# ------------------------------------------------------------

classification_results = (
    results_df[
        results_df["Test_Accuracy"].notna()
    ]
    [
        [
            "Experiment",
            "Type",
            "Classification_supervision",
            "Best_epoch",
            "Test_Accuracy",
            "Test_F1_Macro"
        ]
    ]
    .sort_values(
        "Test_Accuracy",
        ascending=False
    )
)

classification_path = os.path.join(
    RESULTS_DIR,
    "classification_comparison.csv"
)

classification_results.to_csv(
    classification_path,
    index=False
)

print("\n")
print("=" * 100)
print("CLASSIFICATION COMPARISON")
print("=" * 100)

display(
    classification_results
)


# ------------------------------------------------------------
# Segmentation comparison
# ------------------------------------------------------------

segmentation_results = (
    results_df[
        results_df["Test_Dice"].notna()
    ]
    [
        [
            "Experiment",
            "Type",
            "Segmentation_supervision",
            "Best_epoch",
            "Test_Dice",
            "Test_IoU"
        ]
    ]
    .sort_values(
        "Test_Dice",
        ascending=False
    )
)

segmentation_path = os.path.join(
    RESULTS_DIR,
    "segmentation_comparison.csv"
)

segmentation_results.to_csv(
    segmentation_path,
    index=False
)

print("\n")
print("=" * 100)
print("SEGMENTATION COMPARISON")
print("=" * 100)

display(
    segmentation_results
)


# ============================================================
# 44. VISUALIZATION FUNCTION
# ============================================================

def visualize_mtl_prediction(
    model,
    loader,
    title="MTL Prediction"
):

    model.eval()

    images, true_masks, labels = next(
        iter(loader)
    )

    images_gpu = images.to(
        device
    )

    with torch.no_grad():

        pred_masks_logits, pred_classes = (
            model(images_gpu)
        )

    pred_masks = (
        torch.sigmoid(
            pred_masks_logits
        ) > 0.5
    ).float()

    sample_image = (
        images[0]
        .cpu()
        .permute(1, 2, 0)
        .numpy()
    )

    sample_image = (
        sample_image
        * np.array(IMAGENET_STD)
        +
        np.array(IMAGENET_MEAN)
    )

    sample_image = np.clip(
        sample_image,
        0,
        1
    )

    true_mask = (
        true_masks[0]
        .cpu()
        .squeeze()
        .numpy()
    )

    pred_mask = (
        pred_masks[0]
        .cpu()
        .squeeze()
        .numpy()
    )

    true_class = CLASS_NAMES[
        labels[0].item()
    ]

    predicted_class = CLASS_NAMES[
        pred_classes[0]
        .argmax()
        .item()
    ]

    plt.figure(
        figsize=(15, 5)
    )

    plt.subplot(
        1,
        3,
        1
    )

    plt.imshow(
        sample_image
    )

    plt.title(
        f"Original\n"
        f"True: {true_class}"
    )

    plt.axis("off")

    plt.subplot(
        1,
        3,
        2
    )

    plt.imshow(
        true_mask,
        cmap="gray"
    )

    plt.title(
        "Ground Truth Lesion Mask"
    )

    plt.axis("off")

    plt.subplot(
        1,
        3,
        3
    )

    plt.imshow(
        pred_mask,
        cmap="gray"
    )

    plt.title(
        f"Predicted Lesion Mask\n"
        f"Pred: {predicted_class}"
    )

    plt.axis("off")

    plt.suptitle(
        title
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 45. VISUALIZE FULL MTL
# ============================================================

visualize_mtl_prediction(
    all_models[
        "E3_FULL_MTL"
    ],
    test_loader,
    title="E3 — Full-label MTL"
)


# ============================================================
# 46. VISUALIZE 50/50 PARTIAL MTL
# ============================================================

visualize_mtl_prediction(
    all_models[
        "E4_PARTIAL_MTL_50_50"
    ],
    test_loader,
    title="E4 — 50/50 Partial-label MTL"
)


# ============================================================
# 47. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 100)
print("EXPERIMENT COMPLETE")
print("=" * 100)

print(
    f"Total dataset   : {len(full_dataset)}"
)

print(
    f"Train samples   : {len(train_idx)}"
)

print(
    f"Validation      : {len(val_idx)}"
)

print(
    f"Test            : {len(test_idx)}"
)

print(
    f"Image size      : {IMG_SIZE}x{IMG_SIZE}"
)

print(
    f"Batch size      : {BATCH_SIZE}"
)

print(
    f"Epochs          : {EPOCHS}"
)

print(
    f"Learning rate   : {LEARNING_RATE}"
)

print(
    f"Backbone        : ResNet-34"
)

print(
    f"Seed            : {SEED}"
)

print(
    f"Device          : {device}"
)

print("\nFinal results:")
print(results_path)

print("\nSplit file:")
print(split_path)

print("\nSaved model directory:")
print(SAVE_DIR)

print(
    "\n🎉 All STL / Full-MTL / Partial-MTL experiments completed."
)
```
